```python
# Notebook template, the IPS Framework will add a cell before this one
# defining DATA_FILES as a list of state file paths.

# In this example, this notebook is generated during the time loop.

import json

value: dict[int, dict[float, list[str]]] = ips_analysis_api.get_child_data()
for runid, runid_map in value.items():
    print(f'-----------ANALYZING CHLID RUN {runid} ---------------')
    for timestep, data_files in runid_map.items():
        print(f'--- TIMESTEP {timestep} ---')
        for path in data_files:
            with open(path, 'r') as f:
                data = json.loads(f.read())
            # you can do something with the data here, we're just printing it
            print(data)
            print()
        print()
    print()
print(value)
```
