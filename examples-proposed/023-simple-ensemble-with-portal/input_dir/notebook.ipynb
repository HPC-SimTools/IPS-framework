```python
# Notebook template, the IPS Framework will add a cell before this one
# defining the ips_analysis_api variable.

import json

from bokeh.io import output_notebook
from bokeh.plotting import figure, show

output_notebook()

plot = figure(width=1000, height=1000)

COLORS = ['red', 'blue', 'green']
color_idx = 0

# This notebook assumes that all ensembles from this run which save JSON files have a similar data format
value: dict[int, dict[float, list[str]]] = ips_analysis_api.get_child_data_by_ensemble_names()
for _runid, runid_map in value.items():  # noqa: PERF102  # show full example
    for _timestep, data_files in runid_map.items():  # noqa: PERF102
        for path in filter(lambda fname: fname.endswith('.json'), data_files):
            with open(path, 'r') as f:
                data = json.loads(f.read())
            plot.scatter(
                data['x_data'],
                data['y_data'],
                marker='plus',
                size=5,
                color=COLORS[color_idx],
                alpha=0.5,
                legend_label=f'{data["word"]} ({data["base_x"]}, {data["base_y"]})',
            )
            color_idx = (color_idx + 1) % len(COLORS)

plot.legend.location = 'top_left'
plot.legend.click_policy = 'hide'

show(plot)
```
