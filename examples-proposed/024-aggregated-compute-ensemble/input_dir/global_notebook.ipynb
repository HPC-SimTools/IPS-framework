```python
import csv
from pathlib import Path

import pandas as pd
```

```python
# Gather ensemble instance data
ensemble_data = ips_analysis_api.get_child_data()
```

`ensemble_data` will be a python dict within a dict that has a list.  The outermost dict is keyed by the run number, the next dict level by the timestamp, and the list within that dict will be of all the data that was registered for the instance run.

```python
ensemble_data
```

For this example, we're interested in aggregating the instance data found in all the CSV files.  (The JSON files contain the solutions for each instance for which there are separate jupyter notebooks, and so are outside the scope of this notebook.)  So, let's create a pandas dataframe from all the CSV files.

```python
rows = []
for k in ensemble_data.keys():
    csv_path = Path(ensemble_data[k][0.0][1])
    print(f'Reading {csv_path}')
    with csv_path.open('r') as csv_file:
        csv_reader = csv.DictReader(csv_file)
        rows.extend(list(csv_reader))
```

```python
df = pd.DataFrame(rows)
```

```python
df
```

```python
# convert start and end times to floats to make new duration column
df['start'] = df['start'].astype(float)
df['end'] = df['end'].astype(float)
```

```python
df['duration'] = df['end'] - df['start']
```
