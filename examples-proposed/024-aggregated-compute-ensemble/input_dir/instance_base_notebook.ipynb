# Instance base notebook

This notebook replicates for each ensemble instance.

```python
import json
from pathlib import Path

%matplotlib notebook
import matplotlib.pyplot as plt
```

```python
# get the JSON data file generated from the instance
data_files = ips_analysis_api.get_data()
```

```python
# echo what we got for a sanity check
data_files
```

```python
for file in data_files[0.0]:
    if file.endswith('_solution.json'):
        solution_json_file = file
        break
else:
    print('JSON solution file for this ensemble is missing')
```

```python
# Now read the JSON file and extract the data.
with open(solution_json_file, "r") as f:
    solution_data = json.load(f)

x = solution_data["x"]
u = solution_data["u"]

# Now plot the data.
plt.plot(x, u)
plt.xlabel("Position (x)")
plt.ylabel("Temperature (u)")
plt.title("Solution of 1D Heat Equation")
plt.grid(True)
plt.savefig(f"solution.png")
```
